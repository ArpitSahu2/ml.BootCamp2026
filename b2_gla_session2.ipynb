```python
students = {
    "arpit": {"marks": 85, "attendance": 80, "study_hours": 2},
    "ani": {"marks": 65, "attendance": 60, "study_hours": 5},
    "ashutosh": {"marks": 40, "attendance": 90, "study_hours": 0},
    "samridh": {"marks": 15, "attendance": 70, "study_hours": 3},
    "somya": {"marks": 75, "attendance": 76, "study_hours": 1}
}


for name, data in students.items():
    marks = data["marks"]
    attendance = data["attendance"]
    study_hours = data["study_hours"]

    print(f"\n{name}'s Report:")


    if marks > 75:
        print("Performance: Strong")
    elif marks >= 40:
        print("Performance: Needs Improvement")
    elif marks < 20:
        print("Performance: Fail - Repeat Class")


    if attendance < 75:
        print("Warning: Low Attendance")


    if study_hours < 1:
        print("Warning: Needs More Study Time")
```
