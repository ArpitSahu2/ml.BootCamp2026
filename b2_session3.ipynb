```python
students = [
    {"name": "Alice", "study_hours": 6, "attendance": 90, "marks": 80},
    {"name": "Bob", "study_hours": 3, "attendance": 70, "marks": 65},
    {"name": "Charlie", "study_hours": 5, "attendance": 80, "marks": 30},
    {"name": "David", "study_hours": 2, "attendance": 60, "marks": 15},
    {"name": "Eve", "study_hours": 7, "attendance": 95, "marks": 92}
]

student_name_input = input("Enter student name: ")

found_student = None
for student in students:
    if student["name"].lower() == student_name_input.lower():
        found_student = student
        break

if found_student:
    print(f"--- Performance Report for {found_student['name']} ---")
    # Marks performance
    if found_student["marks"] > 75:
        print("Performance: strong")
    elif found_student["marks"] >= 40:
        print("Performance: needs improvement")
    else:
        print("Performance: fail repeat class")

    # Attendance warning
    if found_student["attendance"] < 75:
        print("Warning: low attendance")

    # Study hours warning
    if found_student["study_hours"] < 4:
        print("Warning: needs more study time")
else:
    print(f"Student '{student_name_input}' not found.")
```

```python
# Student Starter Code
transactions = [
    {"transaction_id": "TX101", "amount": 6200, "method": "Credit Card"},
    {"transaction_id": "TX102", "amount": 5500, "method": "Gift Card"},
    {"transaction_id": "TX103", "amount": 1200, "method": "Gift Card"},
    {"transaction_id": "TX104", "amount": 8000, "method": "Gift Card"},
    {"transaction_id": "TX105", "amount": 450, "method": "PayPal"}
]

flagged_transactions = []

# Write your code below to find anomalous patterns

#for and in loop

for tx in transactions:
  if tx["amount"]>5000 and tx["method"]=="Gift Card":
    flagged_transactions.append(tx["transaction_id"])

print("Flagged Transactions:",flagged_transactions)
```
